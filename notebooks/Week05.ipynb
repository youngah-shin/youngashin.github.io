<p><font color="#0F6E56" size="7"><strong>The Basic Structure of an <code>if</code> Statement</strong></font></p>
<ul>
<li><font size="5">An <strong><code>if</code> statement</strong> is the most basic conditional structure.</font></li>
<li><font size="5">The indented code runs only when the condition evaluates to <strong><code>True</code></strong>.</font></li>
</ul>
<div class="highlight"><pre><span></span><span class="k">if</span> <span class="n">condition</span><span class="p">:</span>
    <span class="n">code_to_execute</span>
</pre></div>

```python
age = 20

if age >= 18:
    print('You are an adult.')

print('Program finished.')
```

```python
temperature = 28

if temperature > 30:
    print('It is very hot.')
    print('It may be better to stay indoors.')

print("That's today's weather update.")
```

<p><font color="#0F6E56" size="7"><strong>The <code>if</code>-<code>else</code> Statement</strong></font></p>
<ul>
<li><font size="5">Use <strong><code>else</code></strong> to specify what should happen when the condition evaluates to <strong><code>False</code></strong>.</font></li>
</ul>
<div class="highlight"><pre><span></span><span class="k">if</span> <span class="n">condition</span><span class="p">:</span>
    <span class="n">code_to_run_when_true</span>
<span class="k">else</span><span class="p">:</span>
    <span class="n">code_to_run_when_false</span>
</pre></div>

```python
time = 14  # 24-hour format

if time < 12:
    print('Good morning!')
else:
    print('Good afternoon!')
```

```python
score = 75

if score >= 60:
    print('You passed.')
    print('Congratulations!')
else:
    print('You did not pass.')
    print('Try again next time.')
```

```python
is_raining = False

if is_raining:
    print('Take an umbrella.')
    print('Be careful in the rain!')
else:
    print('The weather is nice.')
    print('Have a great day!')
```

<p><font color="#0F6E56" size="7"><strong>Values Evaluated as <code>True</code> or <code>False</code></strong></font></p>
<ul>
<li><font size="5">When a value is used as a condition, Python can treat it as <code>True</code> or <code>False</code>.</font></li>
<li><font size="5">Common values treated as <strong><code>False</code></strong> include:</font>
<ul>
<li><font size="5"><code>False</code></font></li>
<li><font size="5">the number <code>0</code></font></li>
<li><font size="5">an empty string <code>''</code></font></li>
<li><font size="5">an empty list <code>[]</code></font></li>
<li><font size="5">an empty tuple <code>()</code></font></li>
<li><font size="5">an empty dictionary <code>{}</code></font></li>
<li><font size="5">an empty set <code>set()</code></font></li>
<li><font size="5"><code>None</code></font></li>
</ul></li>
<li><font size="5">Most other values are treated as <strong><code>True</code></strong>.</font></li>
<li><font size="5">You may hear these described as <strong>falsy</strong> and <strong>truthy</strong> values.</font></li>
</ul>

```python
if 0:
    print('This line will not run.')
```

```python
if 1:
    print('This line will run.')
```

```python
empty_list = []

if empty_list:
    print('The list is not empty.')
else:
    print('The list is empty.')
```

<p><font color="#0F6E56" size="7"><strong>The <code>pass</code> Statement</strong></font></p>
<ul>
<li><font size="5">Use <strong><code>pass</code></strong> when Python requires a statement syntactically but you do not want the block to do anything yet.</font></li>

<p><font color="#0F6E56" size="7"><strong>Applications in Data Analysis and AI</strong></font></p>
<ul>
<li><font size="5">Conditional statements are useful in tasks such as <strong>data preprocessing</strong> and simple <strong>model performance checks</strong>.</font>
<ul><li><font size="5"><strong>Data preprocessing</strong> means preparing raw data so it is easier to analyze or use later.</font></li></ul></li>
<li><font size="5">The examples below are intentionally simplified so we can focus on how conditional statements work.</font></li>
</ul>

<ul>
<li><font size="5"><strong>Example:</strong> Simple data preprocessing — clipping a high sensor value</font>
<ul>
<li><font size="5">Suppose <code>100</code> is the chosen upper limit for this simplified example.</font></li>
<li><font size="5">If the sensor value goes above that limit, the program clips it to <code>100</code>.</font></li>
<li><font size="5">This kind of rule can be used during preprocessing when a valid range has been defined for the data.</font></li>
</ul></li>
</ul>

```python
temperature = 185

if temperature > 100:
    temperature = 100
    print('The temperature was above the selected upper limit, so it was adjusted to 100.')
else:
    print('The temperature is within the selected range.')

print('Processed temperature:', temperature)
```

<ul>
<li><font size="5"><strong>Example:</strong> Simple model performance check</font>
<ul>
<li><font size="5">Suppose <code>accuracy</code> is a score from <code>0</code> to <code>1</code> that tells us how often the model was correct in this simplified example.</font></li>
<li><font size="5">The program checks whether the accuracy meets a selected threshold.</font></li>
<li><font size="5">In a real AI project, model evaluation usually considers more than one measure and additional validation.</font></li>
</ul></li>
</ul>

```python
accuracy = 0.85
threshold = 0.8

if accuracy >= threshold:
    print('The model meets the selected accuracy threshold.')
    print('This simplified performance check passes.')
else:
    print('The model does not meet the selected accuracy threshold.')
    print('This simplified performance check does not pass.')
```

<ul>
<li><font size="5">In this section, we will learn how to handle <strong>multiple and more complex conditions</strong>.</font></li>

<p><font color="#0F6E56" size="7"><strong>The <code>if</code>-<code>elif</code>-<code>else</code> Statement</strong></font></p>
<ul>
<li><font size="5">Use <strong><code>if</code>-<code>elif</code>-<code>else</code></strong> to test multiple conditions in sequence.</font></li>
</ul>
<div class="highlight"><pre><span></span><span class="k">if</span> <span class="n">condition1</span><span class="p">:</span>
    <span class="n">code_for_condition1</span>
<span class="k">elif</span> <span class="n">condition2</span><span class="p">:</span>
    <span class="n">code_for_condition2</span>
<span class="k">elif</span> <span class="n">condition3</span><span class="p">:</span>
    <span class="n">code_for_condition3</span>
<span class="k">else</span><span class="p">:</span>
    <span class="n">code_when_all_conditions_are_false</span>
</pre></div>

```python
score = 85

if score >= 90:
    print('Grade A')
elif score >= 80:
    print('Grade B')
elif score >= 70:
    print('Grade C')
else:
    print('Grade D')

print('Grade assignment is complete.')
```

<p><font color="#0F6E56" size="7"><strong>Nested Conditional Statements</strong></font></p>
<ul>
<li><font size="5">A conditional statement can contain another <strong>conditional statement</strong> inside it.</font></li>
</ul>
<div class="highlight"><pre><span></span><span class="k">if</span> <span class="n">outer_condition</span><span class="p">:</span>
    <span class="k">if</span> <span class="n">inner_condition</span><span class="p">:</span>
        <span class="n">code_when_inner_condition_is_true</span>
    <span class="k">else</span><span class="p">:</span>
        <span class="n">code_when_inner_condition_is_false</span>
<span class="k">else</span><span class="p">:</span>
    <span class="n">code_when_outer_condition_is_false</span>
</pre></div>

```python
age = 20
has_license = True

if age >= 18:
    if has_license:
        print('You can drive.')
    else:
        print('You need a driver license.')
else:
    print('You are too young to drive.')
```

<p><font color="#0F6E56" size="7"><strong>Handling More Complex Conditions</strong></font></p>
<ul>
<li><font size="5">You can combine several conditions to handle more complex situations.</font></li>
</ul>

```python
temperature = 25
is_raining = False

if temperature > 30:
    if is_raining:
        print('It is hot and rainy. Stay indoors.')
    else:
        print('It is very hot. Be careful in the sun.')
elif temperature > 20:
    if is_raining:
        print('It is mild but rainy. Take an umbrella.')
    else:
        print('The weather is nice. It is a good day for a walk.')
else:
    if is_raining:
        print('It is chilly and rainy. Dress warmly.')
    else:
        print('It is chilly. Bring a jacket.')
```

<p><font color="#0F6E56" size="7"><strong>Logical Operators</strong></font></p>
<ul>
<li><font size="5">When working with Boolean conditions, use logical operators to combine or reverse conditions.</font></li>
<li><font size="5"><code>and</code>: <code>True</code> when <strong>both</strong> conditions are <code>True</code></font></li>
<li><font size="5"><code>or</code>: <code>True</code> when <strong>at least one</strong> condition is <code>True</code></font></li>
<li><font size="5"><code>not</code>: reverses the Boolean result of a condition</font></li>
</ul>

```python
age = 25
has_license = True
```

```python
if age >= 18 and has_license:
    print('You can drive.')
else:
    print('You cannot drive.')
```

```python
if age < 18 or not has_license:
    print('You cannot drive.')
else:
    print('You can drive.')
```

```python
is_weekend = False
is_holiday = True
```

```python
if is_weekend or is_holiday:
    print('Today is a day off.')
    print('Enjoy your time!')
else:
    print('Today is a weekday.')
    print('Keep up your studies!')
```

<p><font color="#0F6E56" size="7"><strong>The <code>in</code> Operator with Different Data Structures</strong></font></p>
<ul>
<li><font size="5">You can think of <code>in</code> as asking, <strong>“Is this item here?”</strong></font></li>
<li><font size="5">Use <code>in</code> to test membership in a list, tuple, set, or dictionary.</font></li>
<li><font size="5">With a <strong>dictionary</strong>, <code>in</code> checks whether a <strong>key</strong> exists by default.</font></li>
</ul>

<ul>
<li><font size="5"><strong>The <code>in</code> operator with a list</strong></font></li>
</ul>

```python
fruits = ['apple', 'banana', 'orange']

if 'apple' in fruits:
    print('Apple is in the list.')
else:
    print('Apple is not in the list.')
```

<ul>
<li><font size="5"><strong>The <code>in</code> operator with a tuple</strong></font></li>
</ul>

```python
numbers = (1, 2, 3, 4, 5)

if 6 in numbers:
    print('6 is in the tuple.')
else:
    print('6 is not in the tuple.')
```

<p><font size="5"><strong>The <code>in</code> operator with a dictionary</strong></font></p>
<ul><li><font size="5">Remember: for a dictionary, <code>in</code> checks the <strong>keys</strong> by default.</font></li></ul>

```python
person = {'name': 'Alice', 'age': 25, 'city': 'New York'}

if 'age' in person:
    print('Age information is available:', person['age'])
else:
    print('Age information is not available.')
```

<ul>
<li><font size="5"><strong>The <code>in</code> operator with a set</strong></font></li>
</ul>

```python
prime_numbers = {2, 3, 5, 7, 11, 13}

if 4 in prime_numbers:
    print('4 is a prime number.')
else:
    print('4 is not a prime number.')
```

<p><font color="#0F6E56" size="7"><strong>Applications in Data Analysis and AI</strong></font></p>
<ul>
<li><font size="5">Compound conditions and logical operators are useful in <strong>data preprocessing</strong>, <strong>rule-based anomaly checks</strong>, and <strong>result interpretation</strong>.</font></li>
<li><font size="5">Real systems may use more complex methods, but simple rules are useful for learning how conditions work.</font></li>
</ul>

<ul>
<li><font size="5"><strong>Example:</strong> Data preprocessing — detecting and handling an out-of-range value</font><ul>
<li><font size="5">If a value falls outside the valid range, it is replaced with the nearest valid boundary value.</font></li>
<li><font size="5">This helps prepare data in a more suitable form for later analysis or modeling.</font></li>
</ul>
</li>
</ul>

```python
raw_value = 75  # 175, -75
processed_value = 0

if raw_value > 100 or raw_value < 0:
    if raw_value > 100:
        processed_value = 100
    else:
        processed_value = 0
else:
    processed_value = raw_value

processed_value
```

<ul>
<li><font size="5"><strong>Example:</strong> Simple rule-based anomaly check</font>
<ul>
<li><font size="5">This example uses temperature and humidity to demonstrate a few simple rules.</font></li>
<li><font size="5">The thresholds are <strong>example values for learning Python</strong>, not real-world safety limits.</font></li>
<li><font size="5">The program creates an alert message when one of the selected rules is met.</font></li>
</ul></li>
</ul>

```python
temperature = 38.5
humidity = 85

if temperature > 39 or humidity > 90:
    is_abnormal = True
elif temperature > 37.5 and humidity > 70:
    is_abnormal = True
else:
    is_abnormal = False

if is_abnormal:
    alert_message = 'Abnormal condition'
else:
    alert_message = 'Normal condition'

alert_message
```

<p><font color="#0F6E56" size="7"><strong>Conditional Expressions (One-Line <code>if</code>-<code>else</code>)</strong></font></p>
<ul>
<li><font size="5">A <strong>conditional expression</strong> is a compact way to make a simple <code>if</code>-<code>else</code> decision in one line.</font></li>
</ul>
<div class="highlight"><pre>result = value_if_true if condition else value_if_false</pre></div>

```python
x = 10

if x > 0:
    result = 'positive'
else:
    result = 'negative or zero'

result
```

```python
x = 10

result = 'positive' if x > 0 else 'negative or zero'

result
```

<p><font color="#0F6E56" size="7"><strong>Using Conditional Expressions</strong></font></p>

<ul>
<li><font size="5"><strong>Example:</strong> Calculating an absolute value</font></li>
</ul>

```python
x = -5

abs_x = x if x >= 0 else -x

abs_x
```

<ul>
<li><font size="5"><strong>Example:</strong> Checking whether a number is even or odd</font></li>
</ul>

```python
num = 7

parity = 'even' if num % 2 == 0 else 'odd'

parity
```

<p><font color="#0F6E56" size="7"><strong>Advantages and Disadvantages of Conditional Expressions</strong></font></p>
<ul>
<li><font size="5"><strong>Advantages</strong></font><ul>
<li><font size="5">They can make simple code more concise.</font></li>
<li><font size="5">A short condition can sometimes be easier to read when written on one line.</font></li>
</ul>
</li>
<li><font size="5"><strong>Disadvantages</strong></font><ul>
<li><font size="5">Complex conditions can become harder to read.</font></li>

</ul>

</ul>

<p><font color="#0F6E56" size="7"><strong>Writing Clear Conditional Statements</strong></font></p>
<ul>
<li><font size="5">For beginners, the main goal is to make conditional logic <strong>clear and easy to follow</strong>.</font></li>
<li><font size="5">When conditions describe ordered ranges, a top-down structure can make the code easier to read.</font></li>
<li><font size="5">Here, we check the highest grade range first and then move downward.</font></li>
</ul>

```python
score = 85
```

```python
# A clear top-down structure for grade ranges
if score >= 90:
    grade = 'A'
elif score >= 80:
    grade = 'B'
elif score >= 70:
    grade = 'C'
else:
    grade = 'D'

grade
```

<p><font color="#0F6E56" size="7"><strong>Minimizing Nested Conditional Statements</strong></font></p>
<ul>
<li><font size="5">Sometimes the same result can be produced without deeply <strong>nested conditional statements</strong>.</font></li>
<li><font size="5">Reducing unnecessary nesting can improve <strong>readability</strong>.</font></li>
</ul>

```python
x = 10
y = 5
```

```python
if x > 0:
    if y > 0:
        print('Both are positive.')
    else:
        print('Only x is positive.')
else:
    print('x is not positive.')
```

```python
if x > 0 and y > 0:
    print('Both are positive.')
elif x > 0:
    print('Only x is positive.')
else:
    print('x is not positive.')
```

<ul>
<li><font size="5">Excessive nesting can make code harder to maintain and easier to get wrong.</font></li>
<li><font size="5">When possible, keep the structure simple by reducing unnecessary nesting.</font><ul>
<li><font size="5">Simplify or reorganize conditions logically.</font></li>
<li><font size="5">Separate conditions when it makes the logic clearer.</font></li>
<li><font size="5">Use another control structure, such as <code>elif</code>, when appropriate.</font></li>
</ul>
</li>
</ul>

```python
x = 1
y = 2
z = 3
category = ''

if x > 0:
    if y > 0:
        if z > 0:
            category = 'Category A'
        else:
            if x > y:
                category = 'Category B'
            else:
                if y > z:
                    category = 'Category C'
                else:
                    category = 'Category D'
    else:
        if z > 0:
            if x > z:
                category = 'Category E'
            else:
                category = 'Category F'
        else:
            category = 'Category G'
else:
    if y > 0:
        if z > 0:
            category = 'Category H'
        else:
            category = 'Category I'
    else:
        if z > 0:
            category = 'Category J'
        else:
            category = 'Category K'

category
```

<p><font color="#0F6E56" size="7"><strong>Applications in Data Analysis and AI</strong></font></p>
<ul>
<li><font size="5">Conditional expressions can be useful for <strong>data preprocessing</strong> and simple <strong>data transformation</strong> tasks.</font></li>
<li><font size="5">For simple logic, they can make data-processing code more concise while remaining readable.</font></li>
</ul>

<ul>
<li><font size="5"><strong>Example:</strong> Simple scaling</font>
<ul>
<li><font size="5">Suppose the values in this example are known to range from <code>0</code> to <code>max_value</code>.</font></li>
<li><font size="5">Dividing <code>raw_value</code> by <code>max_value</code> scales the value to a range from <code>0</code> to <code>1</code>.</font></li>
<li><font size="5">The conditional expression prevents division by zero when <code>max_value</code> is <code>0</code>.</font></li>
<li><font size="5">Scaling is one common step that may appear when preparing data for later analysis or AI workflows.</font></li>
</ul></li>
</ul>

```python
raw_value = 75
max_value = 100

scaled_value = raw_value / max_value if max_value != 0 else 0

scaled_value
```

<ul>
<li><font size="5"><strong>Example:</strong> Creating a simple binary label</font>
<ul>
<li><font size="5">Sometimes a dataset uses numeric labels such as <code>0</code> and <code>1</code> to represent two categories.</font></li>
<li><font size="5">If a labeling rule is based on a threshold, a conditional expression can create the label.</font></li>
<li><font size="5">In this example, a score above the threshold becomes <code>1</code>; otherwise, it becomes <code>0</code>.</font></li>
<li><font size="5">The conditional expression keeps this simple decision compact.</font></li>
</ul></li>
</ul>

```python
score = 85
threshold = 80

label = 1 if score > threshold else 0

label
```

**Why We Use Python for Data Analysis and AI Modeling**

- A rich library ecosystem:
  - **NumPy**: high-performance numerical computing
  - **Pandas**: data processing and analysis
  - **Matplotlib / Seaborn**: data visualization
- Concise, readable syntax that supports fast development.
- Runs as an interactive mode, which is suitable for processing large volumes of data.
- High compatibility with machine learning and deep learning frameworks.

**Why Python is Popular**

- **Readable Syntax**: concise and intuitive
  - eg., A variable named as 'total_price' rather than 'x'.
- **Versatility**: used in web development, data analysis, artificial intelligence, game development, and many other fields
- **Strong Community**: a vast set of open-source libraries and active support
- **High Productivity**: students accomplish a great deal with little code
- **Industry Demand**: employers ask for Python across many roles, including data scientist and software developer

**`Note`  How to check the popularity of a programming language**

- **TIOBE Index**
  - https://www.tiobe.com/tiobe-index/
  - A widely known indicator of programming language popularity.
  - Limitation: it measures how often a language is mentioned, not how much it is actually used.
- **PYPL (PopularitY of Programming Language)**
  - https://pypl.github.io/PYPL.html
  - An indicator based on Google Trends and search data.
  - Limitation: it measures search frequency, not actual usage.

**Key Features of Google Colab**

- **Code Cell**: a cell in which students write and run Python code
- **Text Cell**: a cell in which students write documentation in Markdown format
- **Change Runtime Type**: select CPU, GPU, or TPU according to the task
- **File Upload / Download**: upload local files, or download files created in Colab

**Create a New Cell**
- Press `Ctrl`+`M` first so that Colab treats the next key as a command rather than text.
- Press `A` to insert a new code cell **above** the selected cell.
- Press `B` to insert a new code cell **below** the selected cell.

**Write Your First Code: Hello, Colab!**

- How to run a cell:
  - Click the run button on the left of the cell.
  - Press `Shift` + `Enter` after pressing `Ctrl`+`M` to use the shortcut.
- Result: `Hello, Colab!` is printed below the cell.

```python
print('Hello, Colab!')
```

**Automatic Output of a Code Cell**

- The result of the **last line** of a code cell is displayed automatically, without a call to `print()`.

```python
a = 10
b = 3

a + b
```

- When a cell contains several expressions, only the last one is displayed automatically.

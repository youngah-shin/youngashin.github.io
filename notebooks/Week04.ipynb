<img src="https://raw.githubusercontent.com/youngkorea/ai-programming-fundamentals/main/img/w04_unit1.svg" width="900" alt="1. Understanding Dictionaries">

- <font size="5">In this section, we will learn about **dictionaries**, another important data structure in Python.</font>
- <font size="5">A dictionary stores data as **"key:value" pairs** and supports efficient lookup by key.</font>

<font color="#0F6E56" size="7">**Creating Dictionaries**</font>

- <font size="5">A **dictionary** is a data structure made of **"key:value" pairs**.</font>
- <font size="5">A dictionary is useful to describe the properties(keys) of something(values).</font>
- <font size="5">A dictionary is known as a **mapping** and it doesn't maintain orders.</font>
- <font size="5">Use curly braces `{}` or `dict()` to create a dictionary, and separate each key from its value with a colon `:`.</font>

```python
empty_dict = {}  # Create an empty dictionary

empty_dict
```

```python
type(empty_dict)
```

```python
empty_dict2 = dict()  # Create an empty dictionary

empty_dict2
```

```python
type(empty_dict2)
```

```python
student = {'name': 'Alice', 'age': 20, 'grade': 'A'}  # Create a simple dictionary with three items

student
```

```python
mixed = {'int': 42, 'float': 3.14, 'string': 'hello', 'list': [1, 2, 3]}  # Store values of different data types

mixed
```

<font color="#0F6E56" size="7">**Accessing and Modifying Dictionaries**</font>

- <font size="5">Use square brackets `[]` with a **key** to fetch or change the associated value.</font>
- <font size="5">The index operation uses the same syntax as that used for sequences(strings, lists, tuples), but the item in `[]` is a key, not a position.</font>

```python
person = {'name': 'John', 'age': 30, 'city': 'New York'}
```

- <font size="5">**Accessing a value**</font>

```python
person['name']
```

```python
person['age']
```

- <font size="5">**Modifying a value**</font>

```python
person['age'] = 31

person['age']
```

```python
person
```

- <font size="5">**Adding a new "key:value" pair**</font>

```python
person['job'] = 'Engineer'

person
```

- <font size="5">**Deleting a "key:value" pair**</font>

```python
del person['city']

person
```

<font color="#0F6E56" size="7">**Characteristics of Dictionary Keys**</font>

- <font size="5">Dictionary keys should use **immutable values**, such as strings, numbers, and tuples of immutable values.</font>
- <font size="5">Mutable objects such as lists cannot be used as dictionary keys.</font>

- <font size="5">**Valid keys**</font>

```python
valid_dict = {
    42: 'answer',
    'pi': 3.14,
    (1, 2): 'tuple key'
}

valid_dict
```

- <font size="5">**Invalid key** — raises a `TypeError`</font>

```python
invalid_dict = {[1, 2]: 'list key'}  # A list cannot be used as a dictionary key
```

<font color="#0F6E56" size="7">**Applications in Data Analysis and AI**</font>

- <font size="5">Dictionaries are widely used in **data analysis and AI** to store and retrieve related information.</font>

- <font size="5">**Example:** A simple text sentiment-score lookup</font>
  - <font size="5">This example retrieves a sentiment score for one word and then adds a new word to the dictionary.</font>
  - <font size="5">It represents a very simple building block of a sentiment-analysis workflow.</font>

```python
sentiment_scores = {
    'good': 1,
    'great': 2,
    'excellent': 2,
    'bad': -1,
    'terrible': -2
}
```

```python
sentiment_scores['good']  # Check the sentiment score of 'good'
```

```python
sentiment_scores['amazing'] = 3  # Add a new word

sentiment_scores
```

<img src="https://raw.githubusercontent.com/youngkorea/ai-programming-fundamentals/main/img/w04_unit2.svg" width="900" alt="2. Working with Dictionaries">

- <font size="5">In this section, we will learn how to use dictionaries more effectively.</font>
- <font size="5">We will work with the `in` operator, dictionary methods, iteration, and key dictionary characteristics.</font>

<font color="#0F6E56" size="7">**The `in` Operator**</font>

- <font size="5">Use the `in` operator to check whether a specific **key** exists in a dictionary.</font>

```python
person = {'name': 'John', 'age': 30, 'city': 'New York'}
```

```python
'name' in person
```

```python
'gender' in person
```

<font color="#0F6E56" size="7">**Dictionary Methods**</font>

- <font size="5">Dictionaries provide several useful **built-in methods**.</font>

```python
person = {'name': 'John', 'age': 30, 'city': 'New York'}
```

- <font size="5">`keys()`</font>
  - <font size="5">Returns all keys in the dictionary.</font>

```python
person.keys()
```

- <font size="5">`values()`</font>
  - <font size="5">Returns all values in the dictionary.</font>

```python
person.values()
```

- <font size="5">`items()`</font>
  - <font size="5">Returns all key-value pairs in the dictionary.</font>

```python
person.items()
```

```python
fruits = {'apple':1, 'banana':10, 'kiwi': 30}
```

```python
for fruit, quantity in fruits.items():
    print(fruit, quantity)
```

- <font size="5">`get()`</font>
  - <font size="5">Retrieves a value by key.</font>
  - <font size="5">If the key exists, `get()` returns its value.</font>
  - <font size="5">If the key **does not exist**, `get()` can return a specified default value instead of raising `KeyError`.</font>

```python
person.get('age', 0)  # Return the value if the key exists
```

```python
person.get('height', 170)  # Return the default value 170 if the key is missing
```

- <font size="5">`update()`</font>
  - <font size="5">Adds or modifies multiple **key-value pairs** at once.</font>

```python
person.update({'job': 'Engineer', 'age': 31})

person
```

<font color="#0F6E56" size="7">**Characteristics of Dictionaries**</font>

- <font size="5">**Insertion order and key-based access**</font>
  - <font size="5">In modern Python, dictionaries **preserve insertion order**.</font>
  - <font size="5">Dictionary equality depends on the **"key:value" pairs**, not on the order in which they were inserted.</font>
  - <font size="5">Dictionaries are accessed by **keys rather than positional indexes**.</font>
  - <font size="5">Dictionaries are designed for efficient **key-based lookup**.</font>

```python
dict1 = {'a': 1, 'b': 2, 'c': 3}
dict2 = {'c': 3, 'a': 1, 'b': 2}
```

```python
dict1
```

```python
dict2
```

```python
dict1 == dict2
```

- <font size="5">**Mutable data type**</font>
  - <font size="5">Dictionaries are **mutable**, meaning that their contents can be modified after creation.</font>

- <font size="5">**Unique keys**</font>
  - <font size="5">Each key in a dictionary must be **unique**.</font>
  - <font size="5">If the same key is written more than once, the later value replaces the earlier one.</font>

```python
dict3 = {'a': 1, 'a': 2}  # The later value overwrites the earlier one
```

```python
dict3
```

```python
dict3['a'] = 3
```

```python
dict3
```

- <font size="5">**Restrictions on key types**</font>
  - <font size="5">Dictionary keys should use **immutable values**.</font>
  - <font size="5">Strings, numbers, and tuples of immutable values can be used as keys.</font>
  - <font size="5">Lists and dictionaries cannot be used as keys.</font>

```python
dict4 = {
    'a': 100,
    1: 200,
    (2, 3): 300
}

dict4
```

```python
dict4 = {
    'a': 100,
    1: 200,
    [2, 3]: 300
}

dict4
```

<font color="#0F6E56" size="7">**Applications in Data Analysis and AI**</font>

- <font size="5">Dictionaries are used to **organize, retrieve, and update data** efficiently in data analysis and AI workflows.</font>

- <font size="5">**Example:** A simple English-to-Spanish word translator</font>

```python
translator = {
    'apple': 'manzana',
    'banana': 'plátano',
    'car': 'coche',
    'dog': 'perro'
}
```

```python
word = 'apple'  # Translate a word
translated_word = translator.get(word, 'translation not found')

word, translated_word
```

```python
translator['cat'] = 'gato'  # Add a new word

translator
```

```python
word = 'computer'  # A word that is not in the dictionary
translated_word = translator.get(word, 'translation not found')

word, translated_word
```

- <font size="5">In this section, we will learn about Python **sets**.</font>
- <font size="5">Sets are neither mappings nor sequences.</font>
- <font size="5">Rather, they are **unordered** collections of **unique** and **immutable** objects.</font>

<font color="#0F6E56" size="7">**Creating Sets**</font>

- <font size="5">Use curly braces `{}` or `set()` to create a set.</font>
- <font size="5">Use `set()` function to create an **empty set** because `{}` creates an empty dictionary.</font>

```python
empty_set = set()  # Create an empty set

empty_set
```

```python
type(empty_set)
```

- <font size="5">**Creating a set with curly braces** `{}`</font>

```python
fruits1 = {'apple', 'banana', 'orange'}

fruits1
```

- <font size="5">**Creating a set with `set()`**</font>

```python
fruits2 = set(['apple', 'banana', 'orange'])

fruits2
```

```python
fruits1 == fruits2
```

```python
type(fruits1), type(fruits2)
```

- <font size="5">**Duplicate elements are removed automatically**</font>

```python
colors = {'red', 'blue', 'green', 'red', 'blue'}

colors
```

<font color="#0F6E56" size="7">**Set Operations**</font>

- <font size="5">Sets support common **mathematical set operations**.</font>

```python
set1 = {1, 2, 3, 4, 5}
set2 = {4, 5, 6, 7, 8}
```

- <font size="5">**Union**</font>

```python
union_set = set1 | set2

union_set
```

- <font size="5">**Intersection**</font>

```python
intersection_set = set1 & set2

intersection_set
```

- <font size="5">**Difference**</font>

```python
difference_set = set1 - set2

difference_set
```

- <font size="5">`len()`</font>
  - <font size="5">Returns the number of elements in a set.</font>

```python
len(difference_set)
```

<font color="#0F6E56" size="7">**Set Methods**</font>

- <font size="5">Sets provide useful methods for adding and removing elements.</font>

```python
fruits = {'apple', 'banana', 'orange'}
```

- <font size="5">`add()`</font>
  - <font size="5">Adds one element to a set.</font>

```python
fruits.add('grape')

fruits
```

- <font size="5">`remove()`</font>
  - <font size="5">Removes a specified element from a set.</font>

```python
fruits.remove('banana')

fruits
```

- <font size="5">`update()`</font>
  - <font size="5">Adds multiple elements to a set at once.</font>

```python
fruits.update(['melon', 'kiwi'])

fruits
```

<font color="#0F6E56" size="7">**The `in` Operator**</font>

- <font size="5">Use the `in` operator to check whether an element is contained in a set.</font>

```python
fruits = {'apple', 'banana', 'orange'}
```

```python
'apple' in fruits
```

```python
'grape' in fruits
```

<font color="#0F6E56" size="7">**Characteristics of Sets**</font>

- <font size="5">**Unordered collection**</font>
  - <font size="5">A set does not guarantee a positional order for its elements.</font>
  - <font size="5">Two sets are equal when they contain the same elements, regardless of display order.</font>
  - <font size="5">Sets do not support positional indexing.</font>
  - <font size="5">Sets are useful when the **presence or uniqueness** of elements matters more than their order.</font>

```python
set1 = {1, 2, 3, 4, 5}
set2 = {5, 4, 3, 2, 1}
```

```python
set1
```

```python
set2
```

```python
set1 == set2
```

- <font size="5">**Mutable data type**</font>
  - <font size="5">Sets are **mutable**, meaning that elements can be added or removed after creation.</font>

- <font size="5">**Unique elements**</font>
  - <font size="5">Every element in a set is **unique**; duplicate values are not retained.</font>

```python
set3 = {1, 2, 2, 3, 3, 3}

set3
```

- <font size="5">**Restrictions on element types**</font>
  - <font size="5">Set elements should use **immutable values**.</font>
  - <font size="5">Strings, numbers, and tuples of immutable values can be set elements.</font>
  - <font size="5">Lists and dictionaries cannot be set elements.</font>

```python
set4 = {1, 'a', (2, 3)}
```

```python
set4 = {1, 'a', [2, 3]}
```

<font color="#0F6E56" size="7">**Applications in Data Analysis and AI**</font>

- <font size="5">Sets are useful for working with **unique values** and for simple **data preprocessing** tasks.</font>

- <font size="5">**Example:** Extracting unique words from a document</font>

```python
document = ['the', 'quick', 'brown', 'fox', 'jumps', 'over', 'the', 'lazy', 'dog']  # 'the quick brown fox jumps over the lazy dog'
unique_words = set(document)
```

```python
unique_words
```

```python
len(unique_words)
```

- <font size="5">**Example:** Removing duplicate data</font>
  - <font size="5">**Note:** Converting data to a set may change the original order.</font>

```python
data = [4, 4, 4, 4, 3, 3, 3, 2, 2, 1]
unique_data = set(data)

unique_data
```

```python
unique_list = list(unique_data)

unique_list
```

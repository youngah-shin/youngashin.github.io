<font color="#0F6E56" size="7">**What is a Program?**</font>

- <font size="5">Last week, we ran this as our first Python **program**.</font>
- <font size="5">A **program** is  sequence of definitions and commands.</font>
- <font size="5">The Python interpreter reads the program, processes its definitions, and executes its commands.</font>
- <font size="5">Now, let's look more closely at what Python programs work with.</font>

```python
print('Hello, Colab!')
```

<font color="#0F6E56" size="7">**Statements (Commands)**</font>

- <font size="5">A **statement (or command)** is an instruction that Python executes.</font>
- <font size="5">Statements tell the Python interpreter what to do.</font>
- <font size="5">The statement below tells the interpreter to call the print() function and display a value.</font>

```python
print('Hello, Colab!')
```

<font color="#0F6E56" size="7">**Objects and Types**</font>

- <font size="5">Programs manipulate objects, and **every object has a type**.</font>
- <font size="5">The type of an object determines **what operations** can be performed on it.</font>
- <font size="5">Use `type()` to check the type of an object.</font>

```python
print(type(5))
print(type(3.0))
print(type(True))
print(type('Hello, Colab!'))
```

<font color="#0F6E56" size="7">**Scalar and Non-Scalar Objects**</font>

- <font size="5">Python objects can be discussed as **scalar** or **non-scalar** according to whether they have an accessible internal structure.</font>
- <font size="5">**Scalar objects** are treated as **indivisible** values in this course.</font>
  - <font size="5">`int` — integers</font>
  - <font size="5">`float` — floating-point numbers</font>
  - <font size="5">`bool` — Boolean values</font>
- <font size="5">**Non-scalar objects** have an **internal structure** that can be accessed.</font>
  - <font size="5">`str` — a sequence of characters</font>
  - <font size="5">`list`, `tuple`, `dict`, and `set`: Learn Later! </font>

<font size="5">We will study lists and tuples in Week 3, and dictionaries and sets in Week 4.</font>

<font color="#0F6E56" size="7">**Operators and Expressions**</font>

- <font size="5">An **operator** specifies an operation to perform on objects or values.</font>
- <font size="5">Objects and operators can be combined to form an **expression**.</font>
- <font size="5">Python evaluates an expression to produce one value.</font>
- <font size="5">**An expression evaluates to a value, and that value has a type.**</font>

<font size="5">A simple expression can be viewed as: `object`  `operator`  `object`</font>

```python
print(3 + 2)
print(type(3 + 2))
```

```python
print(5 / 3)
print(type(5 / 3))
```

<font color="#0F6E56" size="7">**Numeric Types**</font>

- <font size="5">Python provides **numeric types** for representing and working with numbers.</font>
- <font size="5">In this course, we begin with two common numeric types: **`int`** and **`float`**.</font>

<font color="#0F6E56" size="7">**Integers (`int`)**</font>

- <font size="5">An integer is a **whole number without a decimal point**.</font>
- <font size="5">Integers can be **positive, negative, or zero**.</font>
- <font size="5">They are commonly used for counts, repetitions, and numbers of samples.</font>

```python
age = 25

age
```

```python
num_samples = 1000

num_samples
```

```python
temperature = -5

temperature
```

<font color="#0F6E56" size="7">**Floating-Point Numbers (`float`)**</font>

- <font size="5">Floating-point numbers represent values **with a decimal form**.</font>
- <font size="5">They are commonly used for measurements, ratios, probabilities, and model performance values.</font>

```python
height = 175.5

height
```

```python
pi = 3.14159

pi
```

```python
radius = 10
area = pi * (radius ** 2)

area
```

<font color="#0F6E56" size="7">**Operators on `int` and `float`**</font>

- <font size="5">**`+`** Addition</font>
- <font size="5">**`-`** Subtraction</font>
- <font size="5">**`*`** Multiplication</font>
- <font size="5">**`/`** Division &mdash; always produces a **`float`**</font>
- <font size="5">**`//`** Floor division</font>
- <font size="5">**`%`** Remainder</font>
- <font size="5"><b>`**`</b> Exponentiation</font>

```python
10 + 3
```

```python
10 - 3
```

```python
10 * 3
```

```python
10 / 3
```

```python
10 // 3
```

```python
10 % 3
```

```python
2 ** 3
```

<font color="#0F6E56" size="7">**Expression Values and Types**</font>

- <font size="5">**The operator and the types of the objects determine the resulting value and its type.**</font>
- <font size="5">For example, `/` **always produces a floating-point result** in Python.</font>

```python
print(3 + 2, type(3 + 2))
print(5 / 2, type(5 / 2))
```

<font color="#0F6E56" size="7">**Evaluating Expressions**</font>

- <font size="5">Python evaluates an expression **step by step until one value remains**.</font>
- <font size="5">**Parentheses** can be used to control the **order of evaluation**.</font>

```python
without_parentheses = 4 + 2 * 3
with_parentheses = (4 + 2) * 3

without_parentheses, with_parentheses
```

<font color="#0F6E56" size="7">**Data and AI Examples**</font>

- <font size="5">Numeric types appear frequently in data analysis and AI workflows.</font>

- <font size="5">Example: Dataset size</font>

```python
dataset_size = 10000

print('Dataset size:', dataset_size)
```

- <font size="5">Example: Model accuracy</font>

```python
accuracy = 0.9567

print('Model accuracy:', accuracy)
```

<font color="#0F6E56" size="7">**Strings as Non-Scalar Objects**</font>

- <font size="5">A string (`str`) is a **sequence of case-sensitive characters**.</font>
- <font size="5">A string can contain letters, digits, spaces, and special characters.</font>
- <font size="5">Because a string has an internal sequence that can be accessed, it is treated here as a non-scalar object.</font>

<font color="#0F6E56" size="7">**String Literals**</font>

- <font size="5">A string can be written using **single quotes** `'...'` or **double quotes** `"..."`.</font>
- <font size="5">**Triple quotes** can be used for a string that spans multiple lines.</font>

```python
name = 'Alice'

name
```

```python
greeting = "Hello, World!"

greeting
```

```python
multiline = '''This is
a string that spans
multiple lines.'''

print(multiline)
```

<font color="#0F6E56" size="7">**String Operations**</font>

- <font size="5">Strings can be **concatenated** with `+`.</font>
- <font size="5">Strings can be **repeated** with `*`.</font>
- <font size="5">The meaning of an operator depends on the types of its operands.</font>

```python
first_name = 'John'
last_name = 'Doe'

full_name = first_name + ' ' + last_name

full_name
```

```python
echo = 'Hello ' * 3

echo
```

<font color="#0F6E56" size="7">**String Indexing**</font>

- <font size="5">Each character in a string has an **index**.</font>
- <font size="5">**Indexing starts at `0`**. **Negative indexes** count backward from the end of the string.</font>

```python
text = 'Python'

first_char = text[0]
last_char = text[-1]

first_char, last_char
```

<font color="#0F6E56" size="7">**String Slicing**</font>

- <font size="5">**Slicing** extracts part of a string.</font>
- <font size="5">The general form is **`string[start:stop:step]`**.</font>

```python
text[1:4]
```

```python
text[:3]
```

```python
text[-3:]
```

```python
text[::-1]
```

<font color="#0F6E56" size="7">**Data and AI Applications**</font>

- <font size="5">Strings are commonly used for **class labels, model names, version information, messages**, and other textual data.</font>

- <font size="5">Example: Class labels</font>

```python
class_labels = 'cat,dog'

print('Class labels:', class_labels)
```

- <font size="5">Example: Model name and version</font>

```python
model_name = 'Deep Neural Network'
version = 'v2.0'

model_info = model_name + ' ' + version

print('Model information:', model_info)
```

<font color="#0F6E56" size="7">**`bool` as a Scalar Type**</font>

- <font size="5">The Boolean type is **`bool`**.</font>
- <font size="5">It has two Boolean values: **`True`** and **`False`**.</font>
- <font size="5">In Python, **the first letter of `True` and `False` must be uppercase**.</font>

```python
is_student = True

is_student
```

```python
type(True)
```

<font color="#0F6E56" size="7">**Comparison Operators**</font>

- <font size="5">Comparison operators compare values and **produce a Boolean value**.</font>
- <font size="5">**`==`** equal to</font>
- <font size="5">**`!=`** not equal to</font>
- <font size="5">**`<`** less than</font>
- <font size="5">**`>`** greater than</font>
- <font size="5">**`<=`** less than or equal to</font>
- <font size="5">**`>=`** greater than or equal to</font>

```python
x = 5
y = 10

print(x == y)
print(x != y)
print(x < y)
print(x > y)
print(x <= y)
print(x >= y)
```

<font color="#0F6E56" size="7">**Comparing Strings**</font>

- <font size="5">Strings can also be compared.</font>
- <font size="5">Ordering comparisons use **lexicographical ordering**.</font>

```python
name1 = 'Alice'
name2 = 'Bob'

print(name1 == name2)
print(name1 != name2)
print(name1 < name2)
```

<font color="#0F6E56" size="7">**Expression → Value → Type**</font>

- <font size="5">A comparison such as `x < y` is an **expression**.</font>
- <font size="5">The expression evaluates to `True` or `False`, and **the resulting value has type `bool`**.</font>

```python
comparison_result = x < y

print(comparison_result)
print(type(comparison_result))
```

<font color="#0F6E56" size="7">**Logical Operators**</font>

- <font size="5">Logical operators combine Boolean values or Boolean expressions into more complex conditions.</font>
- <font size="5">**`and`** — **`True` only when both conditions are `True`**.</font>
- <font size="5">**`or`** — **`True` when at least one condition is `True`**.</font>
- <font size="5">**`not`** — **reverses** a Boolean value.</font>

```python
a = True
b = False

print(a and b)
print(a or b)
print(not a)
print(not b)
```

<font color="#0F6E56" size="7">**Boolean Expressions**</font>

- <font size="5">Comparison expressions can be combined with logical operators.</font>
- <font size="5">A Boolean expression **evaluates to either `True` or `False`**.</font>

```python
age = 20

is_working_age = age >= 18 and age < 65

is_working_age
```

<font color="#0F6E56" size="7">**Converting Values to Boolean**</font>

- <font size="5">The **`bool()`** function converts other values to Boolean values.</font>
- <font size="5">In general, **`0`, empty strings, empty collections, and `None` convert to `False`**.</font>
- <font size="5">**Most nonzero and non-empty values convert to `True`**.</font>
- <font size="5">Lists and `None` appear here only as examples; they will be discussed in later lessons.</font>

```python
print(bool(0))
print(bool(1))
print(bool(100))
print(bool(-1))
```

```python
print(bool(''))
print(bool('Hello'))
```

```python
print(bool([]))
print(bool([1, 2, 3]))
print(bool(None))
```

<font color="#0F6E56" size="7">**Variables as Names**</font>

- <font size="5">A **variable** is a name used to refer to an object or value in a program.</font>
- <font size="5">A simple box analogy can be helpful at first, but more precisely, a variable name is associated with an object.</font>

```python
x = 5
y = 'Hello'
z = 3.14

x, y, z
```

```python
type(x), type(y), type(z)
```

<font color="#0F6E56" size="7">**Assignment**</font>

- <font size="5">The **`=`** symbol is the **assignment operator**.</font>
- <font size="5">**Python first evaluates the expression on the right-hand side.**.</font>
- <font size="5">The resulting object or value is then **associated with the variable name on the left**.</font>

```python
x = 3 + 2

x
```

<font color="#0F6E56" size="7">**`=` vs. `==`**</font>

- <font size="5">**`=` assigns** a value to a variable name.</font>
- <font size="5">**`==` compares** two values for equality and produces a Boolean value.</font>

```python
x = 5

print(x)
print(x == 5)
```

<font color="#0F6E56" size="7">**Reassignment**</font>

- <font size="5">A variable **can be assigned a new value**.</font>
- <font size="5">**The new value can also have a different type.**.</font>

```python
x = 10
print(x, type(x))

x = 'Python'
print(x, type(x))
```

<font color="#0F6E56" size="7">**Naming Variables**</font>

- <font size="5">Use **meaningful names** that describe the data.</font>
- <font size="5">Variable names may contain letters, digits, and underscores, but **they cannot begin with a digit**.</font>
- <font size="5">Python variable names are **case-sensitive**.</font>
- <font size="5">Use **`snake_case`** for ordinary variable names in this course.</font>

```python
model_accuracy = 0.95
dataset_size = 10000

model_accuracy, dataset_size
```

<font color="#0F6E56" size="7">**Explicit Type Conversion**</font>

- <font size="5">**Type conversion** changes a value from one data type to another.</font>
- <font size="5">Python provides functions such as `int()`, `float()`, `str()`, and `bool()` for explicit conversion.</font>

- <font size="5">Converting a string to an integer</font>

```python
str_num = '123'
int_num = int(str_num)

str_num, type(str_num), int_num, type(int_num)
```

- <font size="5">Converting a string to a floating-point number</font>

```python
str_float = '3.14'
float_num = float(str_float)

str_float, type(str_float), float_num, type(float_num)
```

- <font size="5">Converting numbers to strings</font>

```python
int_num = 456
float_num = 2.718

str_int = str(int_num)
str_float = str(float_num)

str_int, type(str_int), str_float, type(str_float)
```

- <font size="5">Converting an integer to a floating-point number</font>

```python
int_num = 5
float_num = float(int_num)

int_num, type(int_num), float_num, type(float_num)
```

- <font size="5">Converting a floating-point number to an integer</font>

```python
float_num = 7.8
int_num = int(float_num)

float_num, type(float_num), int_num, type(int_num)
```

<font size="5">`int(7.8)` produces `7`. **The decimal part is removed; the value is not rounded to `8`.**.</font>

<font color="#0F6E56" size="7">**Putting the Concepts Together**</font>

```python
raw_accuracy = '0.92'
accuracy = float(raw_accuracy)
passed_threshold = accuracy >= 0.90

print(accuracy)
print(passed_threshold)
```

- <font size="5">`'0.92'` starts as a **string**.</font>
- <font size="5">`float()` converts it to a **floating-point value**.</font>
- <font size="5">`accuracy` is a **variable name** associated with that value.</font>
- <font size="5">`accuracy >= 0.90` is a **Boolean expression**.</font>
- <font size="5">The expression evaluates to the **Boolean value `True`**.</font>
